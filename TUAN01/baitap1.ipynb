 ### Bài 8. Autograd cơ bản    

```python
import torch                                                                                                                                                                       
                                                                                                                                                                                   
x = torch.tensor(                                                                                                                                                                  
    3.6,                                                                                                                                                                           
    requires_grad=True                                                                                                                                                             
)                                                                                                                                                                                  
y = x * x                                                                                                                                                                          
y.backward()                                                                                                                                                                       
                                                                                                                                                                                   
print("x =", x)                                                                                                                                                                    
print("y =", y)                                                                                                                                                                    
print("x.grad =", x.grad)                                                                                                                                                          
                                                                                                                                                                                   
# --- GIẢI THÍCH / YÊU CẦU ---                                                                                                                                                                                                                                                                          
print("\n1. Đạo hàm tính thủ công (2 * 3.6):", 2 * 3.6)                                                                                                                            
print("Gradient tự động bằng Autograd (x.grad):", x.grad.item())                                                                                                                   
# Nhận xét: Kết quả tính tự động bằng Autograd trùng khớp hoàn toàn với phép tính giải tích thủ công.                                                                              
                                                                                                                                                                                   
# 2. Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient:                                                                                                                      
x_new = torch.tensor(3.6, requires_grad=True)                                                                                                                                      
y_new = 3 * x_new**2 + 2 * x_new + 1                                                                                                                                               
y_new.backward()                                                                                                                                                                   
                                                                                                                                                                                   
print("\n2. Đổi hàm y = 3*x^2 + 2*x + 1 tại x = 3.6:")                                                                                                                             
print("Gradient bằng Autograd (x_new.grad):", x_new.grad.item())                                                                                                                                                                                                                                     
print("Kiểm tra tính thủ công (6 * 3.6 + 2):", 6 * 3.6 + 2)              
```

  ### Bài 9. Gradient Descent tìm hệ số đa thức 

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
 X * X,
 X,
 np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
 optimizer.zero_grad()
 y_pred = x @ w
 mse = torch.mean(torch.square(y - y_pred))
 mse.backward()
 optimizer.step()
print("Learned coefficients:")
print(w)                                                                                                                                              
                                                                                                                                                
# 1. So sánh ba hệ số học được với [1, 2, 3]:                                                                                                                                      
print("\n1. So sánh ba hệ số học được:")                                                                                                                                           
print("Hệ số học được:", w.detach().numpy().flatten())                                                                                                                             
print("Hệ số thực tế:  [1, 2, 3]")                                                                                                                                                 
# Nhận xét: Ba hệ số của vector w tiệm cận rất sát với các hệ số ban đầu [1, 2, 3] của đa thức y = 1*x^2 + 2*x + 3.                                                                
                                                                                                                                                                                   
# 2. Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step():                                                                                             
# - optimizer.zero_grad(): Đặt lại gradient của các tham số (w) về 0 trước khi tính toán ở mỗi epoch,                                                                              
#   vì PyTorch mặc định cộng dồn gradient qua các lần gọi backward().                                                                                                              
# - mse.backward(): Tính toán đạo hàm riêng (gradient) của hàm mất mát mse theo từng phần tử của trọng số w bằng cơ chế Autograd.                                                  
# - optimizer.step(): Cập nhật giá trị mới cho các trọng số w dựa trên gradient đã tính và quy tắc tối ưu của thuật toán NAdam.                                                    
                 
```

  ### Bài 10. Giải hệ phương trình bằng Autograd           

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
 y1 = A + B - 9
 y2 = C - D - 1
 y3 = A + C - 8
 y4 = B - D - 2
 sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
 optimizer.zero_grad()
 sqerr.backward()
 optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
                                                                                                                                      
# 1. Kiểm tra nghiệm thu được với bốn phương trình:                                                                                                                                
print("\n1. Kiểm tra lại nghiệm với 4 phương trình:")                                                                                                                              
print(f"A + B = {A.item() + B.item():.4f} (Kỳ vọng: 9)")                                                                                                                           
print(f"C - D = {C.item() - D.item():.4f} (Kỳ vọng: 1)")                                                                                                                           
print(f"A + C = {A.item() + C.item():.4f} (Kỳ vọng: 8)")                                                                                                                           
print(f"B - D = {B.item() - D.item():.4f} (Kỳ vọng: 2)")                                                                                                                           
                                                                                                                                                                                   
# 2. Nhận xét vì sao bài toán có thể có nhiều nghiệm:                                                                                                                              
# - Hệ gồm 4 phương trình nhưng thực chất chỉ có 3 phương trình độc lập tuyến tính:                                                                                                
#   Phương trình 1 trừ Phương trình 3 cộng Phương trình 2:                                                                                                                         
#   (A + B) - (A + C) + (C - D) = B - D = 9 - 8 + 1 = 2 (chính là Phương trình 4).                                                                                                 
# - Do đó ma trận hệ số có hạng rank = 3 < 4 ẩn, hệ phương trình này có vô số nghiệm.                                                                                              
# - Thuật toán Gradient Descent / NAdam sẽ hội tụ về một nghiệm cụ thể phụ thuộc vào giá trị ngẫu nhiên khởi tạo ban đầu. 
```

```python
import torch

# Tạo Tensor X có 24 phần tử
X = torch.arange(24)
print("Tensor X ban đầu:\n", X)

# Reshape lần lượt thành các kích thước yêu cầu
print("\nReshape thành 4 x 6:\n", X.reshape(4, 6))
print("\nReshape thành 2 x 12:\n", X.reshape(2, 12))
print("\nReshape thành 2 x 3 x 4:\n", X.reshape(2, 3, 4))


# 2. Phép toán và thống kê trên hai Tensor $A, B$ kích thước $3 \times 4$

import torch

# Tạo hai Tensor A, B kích thước 3 x 4
A = torch.randn(3, 4)
B = torch.randn(3, 4)

print("Tensor A:\n", A)
print("Tensor B:\n", B)

print("\n1. Tổng A + B:\n", A + B)
print("\n2. Tích phần tử A * B:\n", A * B)
print("\n3. Thương A / B:\n", A / B)
print("\n4. Trung bình theo cột (dim=0):\n", torch.mean(A, dim=0))
print("\n5. Độ lệch chuẩn theo cột (dim=0):\n", torch.std(A, dim=0))

#3. Tính đạo hàm $dy/dx$ cho hàm số $y = 2x^2 + 5x + 3$ tại $x = 2$ bằng Autograd

import torch

x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()

print("Đạo hàm dy/dx tại x = 2 bằng Autograd:", x.grad.item())

#4. Huấn luyện tìm hệ số đa thức $y = 2x^2 - 3x + 5$ (dựa trên cấu trúc Bài 9)

import numpy as np
import torch

# Định nghĩa đa thức chuẩn: 2x^2 - 3x + 5 -> hệ số [2, -3, 5]
polynomial = np.poly1d([2, -3, 5])
N = 20
X_data = np.random.rand(N, 1) * 5
Y_data = polynomial(X_data.squeeze())

# Tạo ma trận đặc trưng XX chứa [x^2, x, 1]
XX = np.hstack([X_data**2, X_data, np.ones_like(X_data)])

# Khởi tạo trọng số w cần học (requires_grad=True)
w = torch.tensor([1.0, 1.0, 1.0], requires_grad=True)
x_tensor = torch.tensor(XX, dtype=torch.float32)
y_tensor = torch.tensor(Y_data, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

# Quá trình huấn luyện (Training loop)
for epoch in range(1500):
    optimizer.zero_grad()
    y_pred = x_tensor @ w
    mse = torch.mean(torch.square(y_pred - y_tensor))
    mse.backward()
    optimizer.step()

print("Các hệ số tìm được sau khi huấn luyện:", w.detach().numpy())
```
